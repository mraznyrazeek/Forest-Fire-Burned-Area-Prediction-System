```python
import joblib
import numpy as np
import pandas as pd
```

```python
model = joblib.load("../models/random_forest_model.pkl")

feature_names = joblib.load("../models/feature_names.pkl")

print("Model loaded successfully!")
print(type(model))
```

```python
print(feature_names)
```

```python
df = pd.read_csv("../../dataset/forestfires_engineered.csv")
```

```python
X = df.drop(
    columns=[
        "area",
        "month",
        "day",
        "season"
    ]
)

X["target"] = np.log1p(df["area"])
X = X.drop(columns=["target"])
```

```python
sample = X.iloc[[0]]

prediction = model.predict(sample)

print("Prediction (log1p):", prediction)

print("Burned Area (ha):", np.expm1(prediction))
```

## Interpretation

The trained Random Forest model and feature names were successfully loaded using Joblib. A sample prediction confirmed that the exported model can be reused independently of the training notebook.

The prediction was generated in the log-transformed scale and converted back to the original burned area (hectares) using the inverse log transformation (`np.expm1()`). This verifies that the trained model is ready to be integrated into the Flask-based web application for real-time forest fire burned area prediction.


## Final Interpretation

The tuned **Random Forest Regressor** achieved a **training R² score of 0.517** and a **testing R² score of -0.029**. Although hyperparameter tuning successfully reduced overfitting compared to the baseline model, the testing performance remained slightly negative, indicating limited predictive ability on unseen data.

This outcome is largely attributed to the characteristics of the **UCI Forest Fires dataset**, which contains only **513 records** (after duplicate removal), a **highly skewed target distribution**, and a **large proportion of zero burned-area observations**. These factors make it difficult for machine learning models to accurately learn and generalize the relationship between the meteorological variables and the burned area.

Despite these limitations, the project successfully demonstrates a complete machine learning workflow, including data preprocessing, feature engineering, model comparison, hyperparameter tuning, performance evaluation, feature importance analysis, and model deployment preparation. The trained Random Forest model was selected as the final model and exported for integration into the Flask-based web application, where it can provide burned-area predictions based on user-provided environmental and weather conditions.

Overall, while the predictive accuracy is constrained by the nature of the dataset, the project successfully meets its objective of developing an end-to-end machine learning-based forest fire burned area prediction system suitable for educational and research purposes.
